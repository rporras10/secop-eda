```python
""" 2.5 Calidad de los contratos con fecha de fin posterior a 2030: ¿la fecha coincide con la duración declarada? """

despues_2030 = secop_df[secop_df['fecha_de_fin_del_contrato'] > limite_superior].copy()

# Duración declarada en días (número x factor de la unidad)
partes_dur = despues_2030['duracion_del_contrato'].str.strip().str.extract(r'^(?P<numero>\d+)?\s*(?P<unidad>.*)$')
factor_a_dias = {'Hora(s)': 1/24, 'Dia(s)': 1, 'Semana(s)': 7, 'Mes(es)': 30, 'Año(s)': 365}
despues_2030['dias_declarados'] = pd.to_numeric(partes_dur['numero'], errors='coerce') * partes_dur['unidad'].map(factor_a_dias)

# Duración según las fechas (desde el inicio; si no hay fecha de inicio, desde la firma)
inicio = despues_2030['fecha_de_inicio_del_contrato'].fillna(despues_2030['fecha_de_firma'])
despues_2030['dias_segun_fechas'] = (despues_2030['fecha_de_fin_del_contrato'] - inicio).dt.days

# Coinciden (±10%): contrato de largo plazo real. No coinciden: la fecha de fin está mal digitada
despues_2030['cuadra_con_duracion'] = (
    (despues_2030['dias_declarados'] - despues_2030['dias_segun_fechas']).abs() <= 0.10 * despues_2030['dias_declarados']
)

print(f"Contratos con fecha de fin después de 2030: {len(despues_2030)}")
print(f"  - Cuadran con su duración declarada (contratos de largo plazo): {despues_2030['cuadra_con_duracion'].sum()}")
print(f"  - No cuadran (fecha de fin mal digitada): {(~despues_2030['cuadra_con_duracion']).sum()}")

# Detalle solo
```

```python
""" 2.7 Análisis de duracion_del_contrato: resumen por unidad y validación por tipo de problema  """

duracion = secop_df['duracion_del_contrato'].str.strip()

# Separar número y unidad
partes = duracion.str.extract(r'^(?P<numero>\d+)?\s*(?P<unidad>.*)$')
numero = pd.to_numeric(partes['numero'], errors='coerce')

# Duración en días 
factor_a_dias = {'Hora(s)': 1/24, 'Dia(s)': 1, 'Semana(s)': 7, 'Mes(es)': 30, 'Año(s)': 365}
dias = numero * partes['unidad'].map(factor_a_dias)
dias_segun_fechas = (secop_df['fecha_de_fin_del_contrato'] - secop_df['fecha_de_inicio_del_contrato']).dt.days

# Resumen por unidad
resumen_unidades = (
    pd.DataFrame({'valor_original': duracion, 'numero': numero, 'unidad': partes['unidad']})
    .groupby('unidad', dropna=False)
    .agg(
        numero_contratos=('valor_original', 'size'),
        sin_numero=('numero', lambda s: s.isna().sum()),
        ejemplo_aleatorio=('valor_original', lambda s: s.sample(1, random_state=42).iloc[0]),
    )
    .sort_values('numero_contratos', ascending=False)
)
resumen_unidades['%'] = (resumen_unidades['numero_contratos'] / len(duracion) * 100).round(2)
display(resumen_unidades)

# Reglas de validación 
es_no_definido = duracion.isna() | (duracion.str.lower() == 'no definido')
condiciones = [
    es_no_definido.fillna(False),
    (~es_no_definido & numero.isna()).fillna(False),
    (numero == 0).fillna(False),
    # Número de 8 dígitos que se puede leer como fecha DDMMAAAA: se digitó una fecha en vez de una duración
    pd.to_datetime(partes['numero'].where(partes['numero'].str.len() == 8), format='%d%m%Y', errors='coerce').notna(),
    (dias > 3650).fillna(False),
    # Tolerancia de 31 días por la aproximación mes = 30 días
    ((dias - dias_segun_fechas).abs() > 31).fillna(False),
]
etiquetas = [
    '1. No definido',
    '2. Unidad sin número',
    '3. Duración cero',
    '4. Fecha digitada como duración',
    '5. Mayor a 10 años',
    '6. No cuadra con fecha de fin - fecha de inicio, mayor a 31 días',
]
problema_duracion = pd.Series(np.select(condiciones, etiquetas, default='OK'), index=secop_df.index)

# Resumen detallado por tipo de problema
resumen_problemas = (
    pd.DataFrame({'etiqueta': problema_duracion, 'valor_original': duracion})
    .groupby('etiqueta')
    .agg(
        numero_contratos=('valor_original', 'size'),
        ejemplo=('valor_original', lambda s: s.sample(1, random_state=42).iloc[0]),
    )
)
resumen_problemas['%'] = (resumen_problemas['numero_contratos'] / len(secop_df) * 100).round(2)
display(resumen_problemas)
```

## 3. Completitud

```python
""" 3.1 Analisis de completitud de los datos de todas las columnas del DataFrame."""

total_instancias = len(secop_df)
print(f"Total de registros: {total_instancias:,}")


def analizar_completitud(df):
    # Nulos y strings vacíos/blancos, son tratados de igual manera como valores incompletos.
    filas = []
    for columna_test in secop_df.columns:
        serie = secop_df[columna_test]
        es_dato_nulo = serie.isnull()
        if serie.dtype == object:
            es_dato_nulo = es_dato_nulo | (serie.astype(str).str.strip() == "")
        nulos = int(es_dato_nulo.sum())
        no_nulos = total_instancias - nulos
        

        filas.append({
            "columna": columna_test,
            "Total Registros": total_instancias,
            "Valores Completos": no_nulos,
            "Valores Nulos o Vacios": nulos,
         
        })

    df_completitud = pd.DataFrame(filas)
    return df_completitud

df_completitud = analizar_completitud(secop_df)


print("\nRESUMEN DE COMPLETITUD")
display(df_completitud)



df_incompletas = df_completitud[df_completitud["Valores Nulos o Vacios"] >= 1].copy()
df_incompletas["% Nulos o Vacios"] = (
    df_incompletas["Valores Nulos o Vacios"] / total_instancias
) * 100


df_incompletas["% Nulos o Vacios"] = df_incompletas["% Nulos o Vacios"].round(4)

print("\nRESUMEN DE COLUMNAS CON NULOS O VACÍOS (>= 1)")
display(df_incompletas)




```

```python
""" 3.4 Consolidado anual de valores pagados, facturados y pendientes de pago. """


# Agrupar por año
df_resumen_anual = (
    secop_df.groupby(secop_df['fecha_de_firma'].dt.year)[
        ['valor_pagado', 'valor_facturado', 'valor_pendiente_de_pago']
    ]
    .sum()
    .reset_index()
    .rename(columns={'fecha_de_firma': 'Año'})
)

# Función: expresar todo en billones de pesos para facilitar la lectura
def formatear_cifras_colombia(valor):
    if pd.isna(valor):
        return valor
    
    # Agrupar por billones de pesos en Colombia (12 ceros = 1 Billon)
    val = valor / 1e12
    sufijo = " B"
    
    # Formateo visual (Punto = miles, Coma = decimales)
    if val.is_integer():
        texto_num = f"{int(val):,}"
    else:
        texto_num = f"{val:,.2f}"
    
    texto_es = texto_num.replace(',', 'X').replace('.', ',').replace('X', '.')
    
    return f"{texto_es}{sufijo}"

# 4. Aplicar formato
for col in ['valor_pagado', 'valor_facturado', 'valor_pendiente_de_pago']:
    df_resumen_anual[col] = df_resumen_anual[col].apply(formatear_cifras_colombia)

display(df_resumen_anual)
```

```python
""" 3.5 Rastreo de un posible outlier, top 5 contratos del 2021 por valor pendiente de pago. """

# Filtro por contratos firmados en 2021
contratos_21 = secop_df[secop_df['fecha_de_firma'].dt.year == 2021]

# Top 5 por valor pendiente de pago
top5_pendiente_21 = contratos_21.nlargest(5, 'valor_pendiente_de_pago')

# Formato de cifras: B = billones, M = millones, misma convención de la celda anterior
def formatear_b_o_m(valor):
    if abs(valor) >= 1e12:
        texto, unidad = f"{valor / 1e12:,.2f}", "B"
    else:
        texto, unidad = f"{valor / 1e6:,.2f}", "M"
    # Formato de punto para miles, coma para decimales
    texto = texto.replace(',', 'X').replace('.', ',').replace('X', '.')
    return f"{texto} {unidad}"

# Copia solo para mostrar, los valores numéricos originales quedan intactos con convención de B o M según el numero de ceros.
top5_mostrar = top5_pendiente_21[['id_contrato', 'nombre_entidad', 'objeto_del_contrato']].copy()
top5_mostrar['valor_del_contrato'] = top5_pendiente_21['valor_del_contrato'].apply(formatear_b_o_m)
top5_mostrar['valor_pendiente_de_pago'] = top5_pendiente_21['valor_pendiente_de_pago'].apply(formatear_b_o_m)

display(top5_mostrar)
```

```python
"""3.7 Análisis de formato de las columnas financieras: valor_pagado, valor_facturado y valor_pendiente_de_pago. """

cols_financieras = ['valor_pagado', 'valor_facturado', 'valor_pendiente_de_pago']

# Función para analizar qué caracteres especiales contiene cada columna
def auditar_formato_texto(df, columnas):
    diagnostico = []
    for col in columnas:
        serie_str = df[col].astype(str).str.strip()
        
        tiene_punto = serie_str.str.contains(r'\.', regex=True).sum()
        tiene_coma = serie_str.str.contains(r',', regex=True).sum()
        tiene_ambos = serie_str.str.contains(r'\.', regex=True) & serie_str.str.contains(r',', regex=True)
        solo_numeros = serie_str.str.match(r'^\d+$').sum()
        
        diagnostico.append({
            "Columna": col,
            "Tiene Punto (.)": tiene_punto,
            "Tiene Coma (,)": tiene_coma,
            "Tiene Ambos (. y ,)": tiene_ambos.sum(),
            "Solo Dígitos (Sin separadores)": solo_numeros,
            "Muestras de Texto": serie_str.unique()[:3].tolist() # Ver 3 ejemplos reales
        })
    return pd.DataFrame(diagnostico)

display(auditar_formato_texto(secop_df, cols_financieras))
```

- Anomalia de 2021, el valor_pendiente_de_pago equivale a 41 veces el Presupuesto General de la Nacion de ese año:  se procede a la eliminacion del contrato CO1.PCCNTR.3102623 en eda_secop

```python
"""3.8 Asimetria y Curtosis de las columnas financieras: valor_pagado, valor_facturado y valor_pendiente_de_pago. """
columnas_evaluadas = ['valor_pagado', 'valor_facturado', 'valor_pendiente_de_pago']

# 1. Cálculo sobre la totalidad de los datos
resumen_estadistico = []

for col in columnas_evaluadas:
    skew_val = secop_df[col].skew()
    kurt_val = secop_df[col].kurt()  # Nota: Pandas calcula la curtosis de exceso (normal = 0)
    
    resumen_estadistico.append({
        "Columna": col,
        "Asimetría (Skewness)": round(skew_val, 4),
        "Curtosis (Kurtosis)": round(kurt_val, 4)
    })

df_métricas = pd.DataFrame(resumen_estadistico)

print("MÉTRICAS SOBRE DATOS COMPLETOS (INCLUYENDO ATÍPICOS)")
display(df_métricas)
```

## 4. Unicidad

```python
""" 4.1 Análisis de unicidad de de id_contrato contra fecha_de_firma, fecha_de_inicio_del_contrato y fecha_de_fin_del_contrato """

total_registros = len(secop_df)
print(f"Total de registros: {total_registros:,}")

# Ajustar a la(s) columna(s) que deberían identificar un registro de forma única
columnas_clave = [
    "id_contrato", "fecha_de_firma", "fecha_de_inicio_del_contrato", "fecha_de_fin_del_contrato"
]


duplicados = (
    secop_df
    .groupby(columnas_clave)
    .size()
    .reset_index(name="cantidad_duplicados")
    .query("cantidad_duplicados > 1")
    .sort_values("cantidad_duplicados", ascending=False)
)

num_duplicados = len(duplicados)
if num_duplicados > 0:
    print(f"Se encontraron {num_duplicados:,} llaves duplicadas:")
    display(duplicados.head(20))

    print("\nEjemplos de registros duplicados:")
    primera_llave = duplicados.iloc[0][columnas_clave]
    filtro = (secop_df[columnas_clave] == primera_llave.values).all(axis=1)
    display(secop_df[filtro])
else:
    columnas_str = " + ".join(columnas_clave)
    print(f"No se encontraron duplicados por la combinacion de  {columnas_str}")
    
```

```python
""" 4.2 Analisis de duplicados de la combinacion de fecha_de_inicio_del_contrato + id_contrato + proveedor_adjudicado. """

columnas_clave = ["fecha_de_inicio_del_contrato", "id_contrato", "proveedor_adjudicado"]

# Contar duplicados
duplicados = (
    secop_df
    .groupby(columnas_clave)
    .size()
    .reset_index(name="cantidad_duplicados")
    .query("cantidad_duplicados > 1")
    .sort_values("cantidad_duplicados", ascending=False)
)

# Resultado
num_duplicados = len(duplicados)
if num_duplicados > 0:
    print(f"Se encontraron {num_duplicados:,} combinaciones duplicadas de {columnas_clave}:")
    display(duplicados.head(20))

    print("\nEjemplo de registros duplicados:")
    primera_llave = duplicados.iloc[0][columnas_clave]
    filtro = (secop_df[columnas_clave] == primera_llave.values).all(axis=1)
    display(secop_df[filtro])
else:
    columnas_str = " + ".join(columnas_clave)
    print(f"La combinación de {columnas_str} es única en todos los registros")
```

### Decisiones
- Ambas combinaciones de llave probadas (id_contrato + fechas; fecha_de_inicio + id_contrato + proveedor_adjudicado) resultaron unicas, sin duplicados. Se confirma que la granularidad del dataset es una fila = un `id_contrato`. No se requiere ningun tratamiento de deduplicacion.
